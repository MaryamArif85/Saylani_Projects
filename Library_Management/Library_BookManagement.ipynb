```python
import csv
import os

FILE_NAME = 'library_data.csv'

def load_data():
    library = []
    issued_records = []
    if os.path.exists(FILE_NAME):
        try:
            with open(FILE_NAME, 'r', newline='', encoding='utf-8') as f:
                reader = csv.DictReader(f)
                for row in reader:
                    row['id'] = int(row['id'])
                    row['qty'] = int(row['qty'])
                    library.append(row)
            print("Data loaded successfully!")
        except Exception as e:
            print(f"Error loading file: {e}")
    return library, issued_records


def save_data(library, issued_records):     # Saves data to file
    try:
        with open(FILE_NAME, 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['id', 'name', 'author', 'qty']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(library)
    except Exception as e:
        print(f"Error saving file: {e}")


library, issued_records = load_data()

# 1. Add Book
def add_book():
    print("\n***** Add Book *****")
    try:
        book_id = int(input("Enter Book ID: "))
        # Check for Duplicate ID
        for book in library:
            if book['id'] == book_id:
                print("Book ID already exists. Please choose a different ID.")
                return
        book_name = input("Enter Book Name: ")
        author = input("Enter Author: ")
        qty = int(input("Enter Quantity: "))

        if qty < 0:
            print("Quantity cannot be negative.")
            return

        new_book = {'id': book_id, 'name': book_name, 'author': author, 'qty': qty}
        library.append(new_book)
        save_data(library, issued_records)
        print("Book added successfully!")

    except ValueError:
        print("Invalid input. Please enter valid numbers for ID and Quantity.")

# 2. Issue Book
def issue_book():
    print("\n***** Issue Book *****")

    # Show available books first
    print("\n***** Available Books *****")
    available_found = False
    for book in library:
        if book['qty'] > 0:
            print(f"ID: {book['id']} | {book['name']} by {book['author']} | Qty: {book['qty']}")
            available_found = True

    if not available_found:
        print("No books available to issue right now.")
        return

    try:
        student_name = input("\nEnter Student Name: ")
        if not student_name.strip():
            print("Student name cannot be empty.")
            return
        book_id = int(input("Enter Book ID to issue: "))

        for book in library:
            if book['id'] == book_id:
                if book['qty'] > 0:
                    book['qty'] -= 1
                    issued_records.append({'student': student_name, 'book_id': book_id, 'book_name': book['name']})
                    save_data(library, issued_records)
                    print(f"Book '{book['name']}' issued to {student_name}")
                    return
                else:
                    print("Book is out of stock!")
                    return

        print("Book ID not found!")

    except ValueError:
        print("Invalid input. Book ID must be a number.")

# 3. Return Book
def return_book():
    print("\n***** Return Book *****")

    # Show all books so user knows what can be returned
    print("\n***** All Books *****")
    for book in library:
        print(f"ID: {book['id']} | {book['name']} by {book['author']} | Qty: {book['qty']}")

    try:
        book_id = int(input("\nEnter Book ID to return: "))

        for book in library:
            if book['id'] == book_id:
                book['qty'] += 1
                save_data(library, issued_records)
                print(f"Book '{book['name']}' returned successfully!")
                return

        print("Book ID not found!")

    except ValueError:
        print("Invalid input. Book ID must be a number.")

# 4. Search Book
def search_book():
    print("\n***** Search Book ******")
    print("1. Search by Name")
    print("2. Search by Author")
    choice = input("Choose: ")

    found = False
    if choice == '1':
        name = input("Enter Book Name: ")
        if not name.strip():
            print("Book name cannot be empty.")
            return
        for book in library:
            if name.lower() in book['name'].lower():
                print(f"Found: ID {book['id']} | {book['name']} by {book['author']} | Qty: {book['qty']}")
                found = True
    elif choice == '2':
        author = input("Enter Author Name: ")
        if not author.strip():
            print("Author name cannot be empty.")
            return
        for book in library:
            if author.lower() in book['author'].lower():
                print(f"Found: ID {book['id']} | {book['name']} by {book['author']} | Qty: {book['qty']}")
                found = True
    else:
        print("Invalid choice.")
        return

    if not found:
        print("No books found!")

# 5. Display Report
def display_report():
    print("\n********** Library Report **********")

    total_books = sum(book['qty'] for book in library)

    print(f"Total books in library: {total_books}")
    print(f"Total issued books: {len(issued_records)}")

    print("\n***** Current Books *****:")
    if not library:
        print("No books in the library.")
    else:
        for book in library:
            print(f"ID {book['id']}: {book['name']} - {book['qty']} available")

    print("\n***** Issued Records ******:")
    if not issued_records:
        print("No books issued yet.")
    else:
        for record in issued_records:
            print(f"{record['student']} issued '{record['book_name']}'")

# Menu
while True:
    print("\n=====**** LIBRARY MANAGEMENT SYSTEM ****=====")
    print("1. Add Book")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Search Book")
    print("5. Show Report")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        add_book()
    elif choice == '2':
        issue_book()
    elif choice == '3':
        return_book()
    elif choice == '4':
        search_book()
    elif choice == '5':
        display_report()
    elif choice == '6':
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Try again.")
```
